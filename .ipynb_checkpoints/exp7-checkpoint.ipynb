```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('input.jpg') 
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
plt.figure(figsize=(10,20))
plt.subplot(121);plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB));plt.title("Input Image")
plt.subplot(122);plt.imshow(gray_image,cmap='gray');plt.title("Gray Image")
plt.show()
```

```python
blur = cv2.blur(gray_image, ksize=(3,3))
plt.imshow(blur,cmap='gray');plt.title("Blurred Image");plt.show()
```

```python
medium = np.median(gray_image)
low=int(max(130,0.7+medium))
high=int(max(230,1/3+medium))
```

```python
edges_grayimage = cv2.Canny(gray_image, low, high)
edges_blurredimage = cv2.Canny(blur,low,high)
```

```python
plt.figure(figsize=(10,20))
plt.subplot(121);plt.imshow(edges_grayimage, cmap='gray');plt.title("Canny Edge Detector (Gray Image)");plt.axis('off')
plt.subplot(122);plt.imshow(edges_blurredimage, cmap='gray');plt.title("Canny Edge Detector (Blur Image)");plt.axis('off')
plt.show()
```

```python
lines = cv2.HoughLinesP(edges_grayimage, 1, np.pi / 180, 100, minLineLength=50, maxLineGap=10)
for line in lines:
    x1, y1, x2, y2 = line[0]  
    cv2.line(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  
plt.title("Result of Hough Using Transform")
plt.axis('off')
```
